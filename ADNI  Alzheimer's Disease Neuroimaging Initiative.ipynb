**Data preprocessing**

```python
# Select baseline features (ending with '_bl') + PTID + DX (label)
baseline_cols = [col for col in data.columns if col.endswith('_bl')] + ['PTID', 'DX']
baseline_df = data[baseline_cols].copy()
```

```python
baseline_df.head(10)
## We took only baseline features
```

#**Handling missing data**

```python
# Handling the missing data
#  Checking missing ratio per column
missing_ratio = baseline_df.isnull().mean().sort_values(ascending=False)
print("Missing Value Ratio (Descending):")
print(missing_ratio)

# Dropping columns with more than 50% missing values
baseline_df = baseline_df.loc[:, missing_ratio <= 0.5]

print("\nAfter Dropping Columns with >50% Missing:")
print("Number of columns:", baseline_df.shape[1])

# Adding back important biomarkers manually if they were dropped
important_biomarkers = ['ABETA_bl', 'TAU_bl', 'PTAU_bl', 'FDG_bl', 'PIB_bl', 'AV45_bl', 'FBB_bl']

for col in important_biomarkers:
    if col not in baseline_df.columns and col in data.columns:
        baseline_df[col] = data[col]
        print(f"Added back important biomarker: {col}")

print("\nFinal Number of Columns after Cleaning:", baseline_df.shape[1])
```

```python
# Manually ensuring these biomarkers are kept
final_biomarkers = ['ABETA_bl', 'TAU_bl', 'PTAU_bl', 'FDG_bl']
baseline_df = baseline_df.loc[:, (baseline_df.isnull().mean() <= 0.5) | baseline_df.columns.isin(final_biomarkers)]

print("Final shape after keeping useful biomarkers:", baseline_df.shape)
```

**Missing value imputation**

```python
# for Numerical features(Continuos) we will use KNN Imputer
# for Categorical features we will use mode imputation
from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder

# Separate numerical and categorical columns
num_cols = baseline_df.select_dtypes(include=['float64', 'int64']).columns
cat_cols = baseline_df.select_dtypes(include=['object', 'category']).columns

# Encode categorical variables temporarily (if any)
label_encoders = {}
for col in cat_cols:
    le = LabelEncoder()
    baseline_df[col] = baseline_df[col].astype(str)  # ensure it's string
    baseline_df[col] = le.fit_transform(baseline_df[col])
    label_encoders[col] = le

# Apply KNN Imputer for numerical data
imputer = KNNImputer(n_neighbors=5)
baseline_df[num_cols] = imputer.fit_transform(baseline_df[num_cols])

```

```python
## Now checking the missing value
updated_missing_ratio = baseline_df.isnull().mean().sort_values(ascending=False)
print("Missing Value Ratio (Descending):")
print(updated_missing_ratio)
```

```python
baseline_df.shape
```

**Removing Outlier**

```python
# For outlier detection and removing , we use the IQR .
# we will keep the value which lies between 25 % quantile and 75 % quantile.

# store only numerical column

numerical_cols = baseline_df.select_dtypes(include=['float64', 'int64']).columns

# copying original dataset to keep a original dataset for further use

new_data = baseline_df.copy()

# Now we will calculate  the first quartile ( Q1) and third quartile (Q3) for each selected column
for col in numerical_cols:
    Q1 = baseline_df[col].quantile(0.25)
    Q3 = baseline_df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # keep data between IQR range
    new_data = new_data[(baseline_df[col] >= lower_bound) & (baseline_df[col] <= upper_bound)]
```

```python
print("this is the shape before outlier removal",baseline_df.shape)
print("this is the shape after outlier removal",new_data.shape)
```

```python
print('Best alpha:', LassoCV.best_params_)
print('Best parameter',LassoCV.best_score_)
```

```python
# we get the best alpha value is 0.1. Now we will again fit lasso with this alpha value
new_lasso = Lasso(alpha=0.1)
new_lasso.fit(x_scaled, y)

# Select best features
lasso_features = x.columns[new_lasso.coef_ != 0]
print("Best features:", lasso_features)
print(lasso_features.to_list())
```

```python
## Now we will Combine lasso_features and rf_features

final_features = set(lasso_features) | set(rf_features)
print("Final features:")
print(final_features)
```

**Hyper Parameter Tuning with Optuna**

```python
!pip install catboost
# Importing the required libraries
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
```

```python
## Using optuna for hyper parameter tuning.
## here we use GradientBoosting,RandomForest,XGBoost,CatBoost for model

def objective(trial):
    classifier_name = trial.suggest_categorical('classifier', ['GradientBoosting','RandomForest', 'XGBoost','CatBoost'])

    if classifier_name == 'GradientBoosting':
        n_estimators = trial.suggest_int('n_estimators', 50, 300)
        learning_rate = trial.suggest_float('learning_rate', 0.01, 0.3, log=True)
        max_depth = trial.suggest_int('max_depth', 3, 20)
        min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)

        model = GradientBoostingClassifier(
            n_estimators=n_estimators,
            learning_rate=learning_rate,
            max_depth=max_depth,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            random_state=42
        )
    elif classifier_name == 'RandomForest': # Corrected indentation here
        # Random Forest hyperparameters
        n_estimators = trial.suggest_int('n_estimators', 50, 300)
        max_depth = trial.suggest_int('max_depth', 3, 20)
        min_samples_split = trial.suggest_int('min_samples_split', 2, 10)
        min_samples_leaf = trial.suggest_int('min_samples_leaf', 1, 10)
        bootstrap = trial.suggest_categorical('bootstrap', [True, False])

        model = RandomForestClassifier(
            n_estimators=n_estimators,
            max_depth=max_depth,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            bootstrap=bootstrap,
            random_state=42
        )

    elif classifier_name == 'XGBoost':
        n_estimators = trial.suggest_int('n_estimators', 50, 300)
        max_depth = trial.suggest_int('max_depth', 3, 15)
        learning_rate = trial.suggest_float('learning_rate', 0.01, 0.3, log=True)
        subsample = trial.suggest_float('subsample', 0.5, 1.0)
        colsample_bytree = trial.suggest_float('colsample_bytree', 0.5, 1.0)

        model = XGBClassifier(
            n_estimators=n_estimators,
            max_depth=max_depth,
            learning_rate=learning_rate,
            subsample=subsample,
            colsample_bytree=colsample_bytree,
            use_label_encoder=False,
            eval_metric='mlogloss',
            random_state=42
        )

    elif classifier_name == 'CatBoost':
        depth = trial.suggest_int("depth", 3, 10)
        learning_rate = trial.suggest_float("learning_rate", 0.01, 0.3, log=True)
        iterations = trial.suggest_int("iterations", 100, 500)

        model = CatBoostClassifier(
            depth=depth,
            learning_rate=learning_rate,
            iterations=iterations,
            verbose=0,
            random_state=42
        )

    score = cross_val_score(model, xtrain, ytrain, cv=5, scoring='accuracy').mean()
    return score
```

```python
!pip install optuna
import optuna
```

```python
# Create a study and optimize it
study = optuna.create_study(direction='maximize')
study.optimize(objective, n_trials=50)
```

```python
# Retrieve the best trial
best_trial = study.best_trial
print("Best trial parameters:", best_trial.params)
print("Best trial accuracy:", best_trial.value)
```

```python
## Find the best hyperparameters for each model
import pandas as pd

#Converting Optuna trials to DataFrame
df_trials = study.trials_dataframe()

# Filtering best trials for each model
gb = df_trials[df_trials['params_classifier'] == 'GradientBoosting']
rf = df_trials[df_trials['params_classifier'] == 'RandomForest']
xgb = df_trials[df_trials['params_classifier'] == 'XGBoost']
cb = df_trials[df_trials['params_classifier'] == 'CatBoost']

# Get best performing trial for each model
best_gb = gb.sort_values('value', ascending=False).iloc[0]
best_rf = rf.sort_values('value', ascending=False).iloc[0]
best_xgb = xgb.sort_values('value', ascending=False).iloc[0]
best_cb = cb.sort_values('value', ascending=False).iloc[0]

#Extract only hyperparameters from each trial
gb_params = {k.replace('params_', ''): v for k, v in best_gb.items() if k.startswith('params_')}
rf_params = {k.replace('params_', ''): v for k, v in best_rf.items() if k.startswith('params_')}
xgb_params = {k.replace('params_', ''): v for k, v in best_xgb.items() if k.startswith('params_')}
cb_params = {k.replace('params_', ''): v for k, v in best_cb.items() if k.startswith('params_')}

#Print best hyperparameters
print(" Best GradientBoosting Parameters:\n", gb_params)
print("\n Best RandomForest Parameters:\n", rf_params)
print("\n Best XGBoost Parameters:\n", xgb_params)
print("\n Best CatBoost Parameters:\n", cb_params)
```

# **Stacking**

```python
## Now we will use stacking for improve the accuracy with the help of our model with those best hyper parameter

from sklearn.linear_model import LogisticRegression
## tuning with their best hyper parameter
from sklearn.ensemble import StackingClassifier
# Gradient Boosting
gb_model = GradientBoostingClassifier(
    n_estimators=105,
    learning_rate=0.04367705372907881,
    max_depth=8,
    min_samples_split=4,
    min_samples_leaf=6,
    random_state=51
)

# Random Forest
rf_model = RandomForestClassifier(
    n_estimators=138,
    max_depth=15,
    min_samples_split=8,
    min_samples_leaf=4,
    bootstrap=True,
    random_state=51
)

# XGBoost
xgb_model = XGBClassifier(
    n_estimators=80,
    max_depth=5,
    learning_rate=0.1878293909738388,
    subsample=0.9902078958016296,
    colsample_bytree=0.5418688503815253,
    use_label_encoder=False,
    eval_metric='mlogloss',
    random_state=51
)

# CatBoost
cb_model = CatBoostClassifier(
    iterations=462,
    depth=10,
    learning_rate=0.126551199514587,
    verbose=0,
    random_state=51
)
```

```python
estimators = [
    ('gb', gb_model),
    ('rf', rf_model),
    ('xgb', xgb_model),
    ('cb', cb_model)
]

stacking_model = StackingClassifier(
    estimators=estimators,
    final_estimator=LogisticRegression(),
    cv=5
)
```

```python
stacking_model.fit(xtrain, ytrain)

from sklearn.metrics import accuracy_score

y_pred = stacking_model.predict(xtest)
acc = accuracy_score(ytest, y_pred)
print(f" Stacking Classifier Accuracy: {acc:.4f}")
```

```python
## Now we are using stacking with the same model and set meta model as gradientboosting model

from sklearn.ensemble import StackingClassifier, GradientBoostingClassifier

# Final estimator (meta-learner)
final_estimator = GradientBoostingClassifier(
    n_estimators=100,
    learning_rate=0.1,
    max_depth=3,
    random_state=42
)

# Define stacking model
stacking_model = StackingClassifier(
    estimators=estimators,
    final_estimator=final_estimator,
    cv=5
)

```

```python
stacking_model.fit(xtrain, ytrain)

y_pred = stacking_model.predict(xtest)
acc = accuracy_score(ytest, y_pred)
print(f" Stacking Classifier Accuracy: {acc:.4f}")
```
